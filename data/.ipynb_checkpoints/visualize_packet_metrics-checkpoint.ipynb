# Packet metrics
Run all cells to visualize `packet_metrics.csv`. All rows are loaded, so trials can have different lengths. Change `csv_path` to inspect another trial.


```python
from pathlib import Path
import csv
import matplotlib.pyplot as plt

csv_path = Path("packet_metrics.csv")
with csv_path.open(newline="") as f:
    rows = list(csv.DictReader(f))
if not rows:
    raise ValueError(f"{csv_path} contains no data rows")

def column(name):
    return [float(row[name]) for row in rows]

time_us = column("Target Time (us)")
time_s = [(t - time_us[0]) / 1_000_000 for t in time_us]
print(f"{len(rows):,} packets | {time_s[-1]:.3f} seconds")
```

```python
fig, axes = plt.subplots(3, 1, figsize=(12, 8), sharex=True)
for ax, name in zip(axes, ["Recv Jitter (us)", "Ready Delta (us)", "Recv Calls"]):
    ax.plot(time_s, column(name), linewidth=0.8)
    ax.set_ylabel(name)
    ax.grid(alpha=0.25)
axes[-1].set_xlabel("Elapsed target time (s)")
fig.suptitle(csv_path.name)
fig.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 3))
ax.hist(column("Recv Jitter (us)"), bins=50, edgecolor="white")
ax.set(xlabel="Receive jitter (µs)", ylabel="Packet count", title="Receive jitter distribution")
fig.tight_layout()
plt.show()
```
